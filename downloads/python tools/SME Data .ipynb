```python
import pandas as pd
import numpy as np
import random
import os

# CONFIGURATION
NUM_RECORDS = 100_000
OUTPUT_PATH = os.path.expanduser(r'C:\Users\Admin\Desktop\My systems\datacsvs\SMESAVE.csv')

# SME Persona Definitions
personas = ['Retail', 'Service', 'Manufacturing']
base_revenue = {'Retail': 5000, 'Service': 7000, 'Manufacturing': 10000}
base_expense = {'Retail': 3000, 'Service': 4000, 'Manufacturing': 6000}

# DATA SIMULATION
data = []
for i in range(NUM_RECORDS):
    month = f'M{(i % 12) + 1}'
    persona = random.choice(personas)
    seasonality = np.sin((i % 12) / 2.0) + 1
    revenue = base_revenue[persona] * seasonality * np.random.uniform(0.9, 1.1)
    expense = base_expense[persona] * np.random.uniform(0.8, 1.2)
    churn_rate = np.random.uniform(0.05, 0.15)
    reinvestment = np.random.choice([0, 500, 1000])

    data.append({
        'Month': month,
        'Persona': persona,
        'Revenue': round(revenue, 2),
        'Expense': round(expense, 2),
        'ChurnRate': round(churn_rate, 2),
        'Reinvestment': reinvestment,
        'OverrideRevenue': np.nan,
        'OverrideExpense': np.nan,
        'Commentary': '',
        'StrategicFlag': '',
        'VerifiedBy': ''
    })

df = pd.DataFrame(data)

# OVERRIDE LOGIC
df['FinalRevenue'] = df.apply(
    lambda row: row['OverrideRevenue'] if pd.notnull(row['OverrideRevenue']) else row['Revenue'], axis=1)
df['FinalExpense'] = df.apply(
    lambda row: row['OverrideExpense'] if pd.notnull(row['OverrideExpense']) else row['Expense'], axis=1)
df['OverrideApplied'] = df.apply(
    lambda row: pd.notnull(row['OverrideRevenue']) or pd.notnull(row['OverrideExpense']), axis=1)

# BUSINESS LOGIC
df['NetProfit'] = df['FinalRevenue'] - df['FinalExpense']
df['GrowthScore'] = (
    (df['NetProfit'] / df['FinalRevenue']) * 0.5 +
    ((1 - df['ChurnRate']) * 0.3) +
    (df['Reinvestment'] / 1000 * 0.2)
).round(2)
df['ROI'] = (df['Reinvestment'] * np.random.uniform(0.8, 1.5)).round(2)
df['ChurnRisk'] = (df['ChurnRate'] * 100).round(1)

# EXPORT TO CSV
df.to_csv(OUTPUT_PATH, index=False)
print(f"✅ Data exported to: {OUTPUT_PATH}")
```
